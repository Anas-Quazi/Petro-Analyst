```python

#^ data handling, scaling and accuracy
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import accuracy_score
import pandas as pd

#? supervised machine learning (classification)
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression

#* supervised machine learning (regression)
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

#& unsupervised machine learning 
from sklearn.cluster import k_means
from sklearn.ensemble import IsolationForest
from sklearn.metrics import silhouette_score
import numpy as np
```

```python
df2 = pd.read_csv(r'C:\Users\anasq\OneDrive\Desktop\PetroAnalyst\Data_Preprocessing\EDA_data2.csv')

df4 = pd.read_csv(r'C:\Users\anasq\OneDrive\Desktop\PetroAnalyst\Data_Preprocessing\EDA_data.csv')
```

```python

#? supervised (classifier)

classifiers = {
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest Classifier": RandomForestClassifier(),
    "KNN": KNeighborsClassifier(),
    "Logistic Regression": LogisticRegression()
}

y_classification = df4['LPG_price'].astype(int)


X_r2 = df2
numeric_cols2 = X_r2.select_dtypes(include='number').columns
numeric_cols2 = numeric_cols2.drop('LPG_price') if 'LPG_price' in numeric_cols2 else numeric_cols2
X_scaled2 = MinMaxScaler().fit_transform(X_r2[numeric_cols2])

X_r4 = df4
numeric_cols4 = X_r4.select_dtypes(include='number').columns
numeric_cols4 = numeric_cols4.drop('LPG_price') if 'LPG_price' in numeric_cols4 else numeric_cols4
X_scaled4 = MinMaxScaler().fit_transform(X_r4[numeric_cols4])

classifier_results = []

for name, model in classifiers.items():

    #~ R2 accuracy
    model.fit(X_scaled2, y_classification)
    accuracy2 = accuracy_score(y_classification, model.predict(X_scaled2))

    #~ R4 accuracy
    model.fit(X_scaled4, y_classification)
    accuracy4 = accuracy_score(y_classification, model.predict(X_scaled4))

    diff = abs(accuracy4 - accuracy2) * 100

    classifier_results.append({
        "Classifier Model": name,
        "Round 2 Accuracy": f"{accuracy2 * 100:.2f}%",
        "Round 4 Accuracy": f"{accuracy4 * 100:.2f}%",
        "Difference": f"{diff:.2f}%"
    })

print("--- CLASSIFIER METRICS ---")
print(pd.DataFrame(classifier_results))
```
